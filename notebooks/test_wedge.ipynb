```python
import numpy as np
import sys
sys.path.append("..")
from simple_henry.simulation import create_s_shaped_wedge
```

```python
import matplotlib.pyplot as plt

count = 0
for x_toe in [-0.5, 0.0, 0.5]:
    for x_top in [1.0, 1.5, 2.0, 2.5]:
        for trans_width in [0.001, 0.01, 0.05, 0.1]:
            count += 1
            conc = create_s_shaped_wedge(x_toe=x_toe, x_top=x_top, trans_width=trans_width)
            plt.imshow(conc)
            plt.colorbar(shrink=0.5, aspect=20, pad=0.04)
            plt.title(f"x_toe={x_toe}, x_top={x_top}, trans_width={trans_width}")
            plt.show()

print(f"Total number of combinations: {count}")
```
